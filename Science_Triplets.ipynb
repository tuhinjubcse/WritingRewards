```python
from arxiv_scraper import extract_paper_content
from llms import generate

papers = [
    {"paper_name": "art_or_artifice", "paper_url": "https://arxiv.org/html/2309.14556v3"}
]

writing_instruction = """You are a science writer. You are given the introduction of a technical paper.
Your goal is to write a 150-200 word high-level summary of the paper and why it is important for society.

You should avoid technical language, and make ensure that your writing is engaging and accessible for a large audience.
Try your best to be original and creative in your writing to, which avoiding cliches, awkward words and overused tropes. Do not use ornamental language and focus on nuance, simplicity and clarity.

Paper Introduction:
[[INTRODUCTION]]

Only output the summary in your response, and nothing else.
"""

for paper in papers:
    full_content, intro_content = extract_paper_content(paper["paper_url"])
    response = generate([{"role": "user", "content": writing_instruction}], variables={"INTRODUCTION": intro_content}, model="t-gpt-4o")
    print(response)
```
